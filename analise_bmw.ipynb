```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Carregar o dataset
df = pd.read_csv('dados/base_bmww.csv')

# Exibir as primeiras linhas do dataset
print("Primeiras 5 linhas do dataset:")
print(df.head())

# Informações gerais sobre o dataset
print("
Informações gerais sobre o dataset:")
print(df.info())

# Estatísticas descritivas
print("
Estatísticas descritivas:")
print(df.describe(include='all'))

# Verificar valores ausentes
print("
Valores ausentes por coluna:")
print(df.isnull().sum())

# Análise de Evolução e Tendências Estratégicas (Etapa 2 do PDF)

# 1. Evolução do Volume de Vendas ao longo dos anos
sales_by_year = df.groupby('Year')['Sales_Volume'].sum().reset_index()
plt.figure(figsize=(12, 6))
sns.lineplot(x='Year', y='Sales_Volume', data=sales_by_year, marker='o')
plt.title('Evolução do Volume de Vendas da BMW ao Longo dos Anos')
plt.xlabel('Ano')
plt.ylabel('Volume de Vendas')
plt.grid(True)
plt.show()

# 2. Evolução da Receita (Price_USD * Sales_Volume) ao longo dos anos
df['Revenue'] = df['Price_USD'] * df['Sales_Volume']
revenue_by_year = df.groupby('Year')['Revenue'].sum().reset_index()
plt.figure(figsize=(12, 6))
sns.lineplot(x='Year', y='Revenue', data=revenue_by_year, marker='o')
plt.title('Evolução da Receita da BMW ao Longo dos Anos')
plt.xlabel('Ano')
plt.ylabel('Receita (USD)')
plt.ticklabel_format(style='plain', axis='y') # Evitar notação científica
plt.grid(True)
plt.show()

# 3. Evolução do Preço Médio por Veículo ao longo dos anos
avg_price_by_year = df.groupby('Year')['Price_USD'].mean().reset_index()
plt.figure(figsize=(12, 6))
sns.lineplot(x='Year', y='Price_USD', data=avg_price_by_year, marker='o')
plt.title('Evolução do Preço Médio por Veículo da BMW ao Longo dos Anos')
plt.xlabel('Ano')
plt.ylabel('Preço Médio (USD)')
plt.grid(True)
plt.show()

# 4. Vendas por Tipo de Combustível ao longo dos anos (foco em eletrificados vs combustão)
fuel_type_sales = df.groupby(['Year', 'Fuel_Type'])['Sales_Volume'].sum().reset_index()
plt.figure(figsize=(14, 7))
sns.lineplot(x='Year', y='Sales_Volume', hue='Fuel_Type', data=fuel_type_sales, marker='o')
plt.title('Evolução do Volume de Vendas por Tipo de Combustível')
plt.xlabel('Ano')
plt.ylabel('Volume de Vendas')
plt.legend(title='Tipo de Combustível')
plt.grid(True)
plt.show()

# 5. Comparativo de Vendas por Região ao longo dos anos
region_sales = df.groupby(['Year', 'Region'])['Sales_Volume'].sum().reset_index()
plt.figure(figsize=(14, 7))
sns.lineplot(x='Year', y='Sales_Volume', hue='Region', data=region_sales, marker='o')
plt.title('Comparativo de Volume de Vendas por Região')
plt.xlabel('Ano')
plt.ylabel('Volume de Vendas')
plt.legend(title='Região')
plt.grid(True)
plt.show()

# Análise de Distribuição, Relações, Quantidades e Proporções (Etapa 3 do PDF)

# 6. Modelos que Impulsionam as Vendas na América do Sul
south_america_sales = df[df['Region'] == 'South America']
model_sales_sa = south_america_sales.groupby('Model')['Sales_Volume'].sum().sort_values(ascending=False).reset_index()
plt.figure(figsize=(12, 7))
sns.barplot(x='Sales_Volume', y='Model', data=model_sales_sa.head(10), palette='viridis')
plt.title('Top 10 Modelos Mais Vendidos na América do Sul')
plt.xlabel('Volume de Vendas')
plt.ylabel('Modelo')
plt.show()

# 7. Distribuição de Preços por Tipo de Combustível
plt.figure(figsize=(12, 7))
sns.boxplot(x='Fuel_Type', y='Price_USD', data=df, palette='coolwarm')
plt.title('Distribuição de Preços por Tipo de Combustível')
plt.xlabel('Tipo de Combustível')
plt.ylabel('Preço (USD)')
plt.grid(axis='y')
plt.show()

# 8. Relação entre Quilometragem e Preço
plt.figure(figsize=(12, 7))
sns.scatterplot(x='Mileage_KM', y='Price_USD', hue='Fuel_Type', data=df, alpha=0.6)
plt.title('Relação entre Quilometragem e Preço')
plt.xlabel('Quilometragem (KM)')
plt.ylabel('Preço (USD)')
plt.grid(True)
plt.show()

# 9. Proporção de Vendas por Classificação (High/Low)
sales_classification_counts = df['Sales_Classification'].value_counts().reset_index()
sales_classification_counts.columns = ['Sales_Classification', 'Count']
plt.figure(figsize=(8, 8))
plt.pie(sales_classification_counts['Count'], labels=sales_classification_counts['Sales_Classification'], autopct='%1.1f%%', startangle=90, colors=sns.color_palette('pastel'))
plt.title('Proporção de Vendas por Classificação')
plt.axis('equal') # Garante que o gráfico de pizza seja circular.
plt.show()
```
